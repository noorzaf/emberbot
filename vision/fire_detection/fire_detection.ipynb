# EmberBot — fire detection

EmberBot's Mi Camera 2K streams to its app on the laptop. This notebook screen-captures that
live feed (a 640×600 region at the top-left of the screen) with `mss`, runs an OpenCV Haar
cascade on each frame, and draws a red box labelled **FIRE DETECTED!** around any match.

Press **q** in the video window to stop.

```python
import cv2
import numpy as np
import mss
```

```python
monitor = {
    "top": 40,
    "left": 0,
    "width": 640,
    "height": 600
}
```

```python
sct = mss.mss()
```

```python

# Load fire detection Haar cascade model
fire_cascade = cv2.CascadeClassifier("fire_detection_cascade_model.xml")

# %%
while True:
    # Capture screen region
    frame = np.array(sct.grab(monitor))
    
    # Convert to BGR (removes alpha channel)
    frame = cv2.cvtColor(frame, cv2.COLOR_BGRA2BGR)
    
    # Convert to grayscale for detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Detect fire regions in the frame
    fires = fire_cascade.detectMultiScale(gray, 1.2, 5)

    # Draw rectangles around detected fire areas
    for (x, y, w, h) in fires:
        cv2.rectangle(frame, (x-20, y-20), (x+w+20, y+h+20), (0, 0, 255), 2)
        cv2.putText(frame, "FIRE DETECTED!", (x, y - 10), 
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)

    # Display the screen feed
    cv2.imshow("Home Feed", frame)

    # Exit on 'q' key press
    if cv2.waitKey(1) == ord('q'):
        break

cv2.destroyAllWindows()
```
